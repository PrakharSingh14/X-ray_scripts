# SoLEXS data download

Downloads SoLEXS day files from PRADAN for the days in `days.txt` and unzips them into `Data/YYYYMMDD/{SDD1,SDD2}/`.

**Before running:** put a fresh PRADAN `solexs_20*.py` script (or `cookie.txt`) in this folder, and list your days in `days.txt`.
Details: `README.md`.

## Settings

```python
DAYS_FILE  = 'days.txt'   # .txt or .csv with your days
DAYS       = None         # or a list, e.g. ['2025-09-09', '2025-01-01 : 2025-01-03']
OUTPUT_DIR = '.'          # where Data_zip/ and Data/ go
KEEP_ZIPS  = True         # False = delete zips after unzipping
DRY_RUN    = False        # True = only show what would be done
```

## Run

```python
import importlib
import solexs_download as sd
importlib.reload(sd)

status = sd.run(days=DAYS, days_file=DAYS_FILE, output_dir=OUTPUT_DIR,
                keep_zips=KEEP_ZIPS, dry_run=DRY_RUN)
```

## Check one day

```python
data_dir = sd._resolve(OUTPUT_DIR) / 'Data'
example = next((data_dir / d for d in sorted(status) if sd.is_done(data_dir / d)), None)
if example is None:
    print('Nothing unzipped yet')
else:
    print(example)
    for p in sorted(example.rglob('*')):
        if p.is_file() and not p.name.startswith('.'):
            print(f'  {str(p.relative_to(example)):60s} {p.stat().st_size/1024**2:8.2f} MB')
```
